# What is T-sne? 
t-sne stands for t distributed stocahistic neighbour embedding . <br>
it is a statistical techinque for visualising high dimensional data by reducing it to lower dimensional space <br><br>

## Why t-sne if we already have pca?
1) t-sne work good with non linear data than pca<br>
2) t-sne is good at visualisation <br>

## How t-sne works?
t-sne try to preserve local structure of data instead of global structure <br>
pca capture global structure (by varaince)<br>
