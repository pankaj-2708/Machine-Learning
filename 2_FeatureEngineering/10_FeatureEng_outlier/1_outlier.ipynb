# Local Outlier Factor (LOF) Algorithm

The Local Outlier Factor (LOF) is used to identify anomalous points in a dataset by comparing the local density of each point with that of its neighbors.

---

## Step 1: Calculate k-distance for each point
For every point **p** in the dataset:
- Calculate the distance from **p** to every other point.
- Determine the *k-th smallest distance*; this is the **k-distance** for point **p**.

---

## Step 2: Compute reachability distance
For every point **p** and each of its k-nearest neighbors **o**:
- The reachability distance from **p** to **o** is defined as:

reachability_distance_k(p, o) = max( k-distance(o), d(p, o) )

where:
- d(p, o) = actual distance between **p** and **o**.

---

## Step 3: Calculate Local Reachability Density (LRD)
For each point **p**:
- The Local Reachability Density (LRD) of **p** is the inverse of the average reachability distance from **p** to its k-nearest neighbors:

LRD_k(p) = 1 / ( ( Σ reachability_distance_k(p, o) ) / k)

---

## Step 4: Compute the Local Outlier Factor (LOF)
For each point **p**:
- The Local Outlier Factor (LOF) is computed by averaging the ratio of the LRD of **p**’s neighbors to the LRD of **p**:

LOF_k(p) =  Σ ( LRD_k(o) / k )/ LRD_k(p) 

---

## Interpretation
- If **LOF ≈ 1** → point **p** has a density similar to its neighbors (normal point).
- If **LOF > 1** → point **p** has a lower density than its neighbors (possible outlier).
- The higher the LOF, the stronger the indication that **p** is an outlier.
