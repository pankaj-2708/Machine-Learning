# Hard SVM

in this we assume our data is ideal (i.e perfectly lineaarly seprable)

<img src="https://upload.wikimedia.org/wikipedia/commons/thumb/7/72/SVM_margin.png/330px-SVM_margin.png">

here w.x-b=0 is that perfect line 

and our descision rule is w.x+b>0 means +ve category
and our descision rule is w.x+b<0 means -ve category

or we can say in sort descison rule = sign(w.x+b)


and w.x-b=1 is the line above which points are of one category 
and w.x-b=-1 is the line above which points are of second category 

instead of 1 and -1 numbers can be anything like 45 and -45 both of equal magnitude but we have taken 1 and -1 for convinenet and their magnitude is same because they are at equal distance from center line 

for getting best line our goal is to find out that distance b/w  w.x-b=-1 and  w.x-b=1 and maximise it 

and the points that touches w.x-b=+-1 are known as Support vectors
