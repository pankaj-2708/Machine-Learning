# core idea of svr 

To find a function that approximates the target values as closely as possible, while being as simple (flat) as possible, and ignoring small errors within a margin of tolerance (ε).

we try to find simple function beacuse more complex function overfit and we do this by minimizing norm of w or by minimising norm of  w**2 /2 (for mathematical convienet) 

we ignore small errors by assinging penalty to terms which are outside the margin of tolerence
