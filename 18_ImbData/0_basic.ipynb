# Handling imbalanced data

## ways of handling imbalanced data

### 1- Undersampling
        in this we decrease the no of rows that are in majority
        from imblearn.under_sampling import RandomUnderSampler

### 2- Oversampling
        increase the no of rows that are in minority by creating duplicates
        from imblearn.over_sampling import RandomOverSampler

### 3- SMOTE
        For each sample x in the minority class:
        Find k nearest neighbors (usually k=5) from the same class.
        Randomly select one or more of these neighbors, say x'.
        Create a synthetic point using the formula:
            synthetic = x + rand(0,1) * (x' - x)

        Disadvantages-
            don't handle categorical variable
            computationally expensive
            depend's on neighours neighbours
            sensitive to outliers
            balanced achived may not represent true nature
