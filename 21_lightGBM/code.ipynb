```python
import numpy as np
import pandas as pd
from lightgbm import LGBMRegressor,LGBMClassifier,plot_importance
from sklearn.datasets import make_regression
from sklearn.model_selection import cross_val_score 
from sklearn.datasets import make_classification
```

# Regression

```python
X,y=make_regression(n_features=2,n_targets=1,n_informative=2,n_samples=100000)
lgb=LGBMRegressor(n_estimators=10,max_depth=5,subsample=0.8,  boosting_type='goss'  ,        
    colsample_bytree=0.8,    
    device='gpu',
    reg_alpha=0.0,
    reg_lambda=0.0)
score=cross_val_score(lgb,X,y)
score.mean(),score.max()
```

# Clasification

```python
X,y=make_classification(n_features=5,n_classes=2,n_informative=2,n_samples=100000)
lgb=LGBMClassifier(n_estimators=10,max_depth=5,subsample=0.8,  boosting_type='goss'  ,        
    colsample_bytree=0.8,    
    device='gpu',
    reg_alpha=0.0,
    reg_lambda=0.0)
score=cross_val_score(lgb,X,y)
score.mean(),score.max()
```
