# AdaBoost vs Gradient Boosting vs XGBoost vs LightGBM

## AdaBoost
**Core idea:** Reweight samples.

- Trains trees directly on the target values.
- After each tree:
  - Misclassified samples get higher weights.
  - Correctly classified samples get lower weights.
- Sample weights influence the next tree during training.
- Final prediction:
  $\hat{y} = \sum_t \alpha_t f_t(x)$
  where $$(\alpha_t)$$ depends on the tree's error.

**Key characteristics**
- No gradients or Hessians
- Typically uses exponential loss
- Focuses strongly on hard-to-classify samples

---

## Gradient Boosting (GBDT)
**Core idea:** Fit pseudo-residuals.

- Each new tree is trained on the **negative gradient of the loss**.
- For MSE loss, pseudo-residuals = true residuals.
- For other losses, pseudo-residuals differ from raw residuals.
- Model update:
  $F_t(x) = F_{t-1}(x) + \eta f_t(x)$

**Key characteristics**
- Uses first-order gradients
- General boosting framework
- Sequential and relatively slow

---

## XGBoost
**Core idea:** Second-order gradient boosting with strong regularization and system optimizations.

Enhancements over GBDT:
- Uses **gradients and Hessians** (2nd-order Taylor expansion).
- Split selection based on **gain**:
  $\text{Gain} = \frac{G_L^2}{H_L + \lambda} + \frac{G_R^2}{H_R + \lambda} - \frac{G^2}{H + \lambda}$
- Explicit regularization (L1, L2).
- Handles missing values automatically.
- Parallelized split finding.
- Supports row and column subsampling.

**Key characteristics**
- More stable and accurate than vanilla GBDT
- Slower than LightGBM on very large datasets

---

## LightGBM
**Core idea:** Faster gradient boosting through smarter tree growth and sampling.

Enhancements over XGBoost:
- **Leaf-wise (best-first) tree growth** instead of level-wise.
- **GOSS (Gradient-based One-Side Sampling)**:
  - Keep samples with large gradients.
  - Randomly sample from small-gradient samples.
  - Sampling is done **before each tree**, after gradient computation.
- **EFB (Exclusive Feature Bundling)** to reduce feature dimensionality.
- Histogram-based split finding.

**Key characteristics**
- Extremely fast on large datasets
- Can overfit if tree depth is not controlled

---

## One-line Summary
- **AdaBoost** → reweight samples  
- **Gradient Boosting** → fit pseudo-residuals  
- **XGBoost** → gradient boosting + second-order optimization + regularization  
- **LightGBM** → XGBoost-style boosting + gradient-based sampling + leaf-wise growth
