# light gbm stands for Light Gradient Boosting Machine
it was developed by Microsoft<br>
Light gbm is made for huge datasets <br>
## Core features of Light gbm
1) up to 20 time faster than xgboost and more memory efficient that it <br>
2) acuuracy and other metrics are similar to xgboost<br>
3) works with categorical data (without encoding)<br>
4) it is available in multiple languages <br>
5) Supports distributed computing<br>
6) Supports GPU<br>
7) can handle missing and sparse data<br>
8) can work with multiple loss fn<br>

### Light gbm and xgboost have same objective function
### Light gbm also uses approx greed method for tree building 
Light gbm uses these three techniques which are different from xgboost <br>
1) Best fit tree (for accuracy and other metrics)<br>
2) GOSS (for speed and memory efficeny)<br>
3) EFB (for speed and memory efficeny)<br>

# Tree building technique 
## Depth first
this is the most commonly used it is used in descisson tree and xgboost what it do is built tree depthwise firsly create root node then its left node
then left node of left nodeand so on checkout descion tree customised class

## Breadth first / level first
in this we create tree in level wise manner firstly all tree at a level are created before going to next level among tree's at same level the one which give best gain is splitted first 

## Leaf first / Best fit tree 
in this at each time we check all the nodes that are yet to be split and split one which has maximum gain <br>
used by light gbm <br>

### Charectestics of leaf tree
1) it minimises the lost better than above two/can find more accurate trees <br>
2) It creates unbalanced tree<br>
3) Computational saving - by focousing on most promising and fewer splits than other strategy<br>
4) can lead to overfitting<br>

xgboost has also implemented this method of spliiting tree we can control it using grow_policy hperparameter lossguide for best fit strategy

# GOSS 
Full form of GOSS is gradient based one sided sampling<br>
Goss reduces the no of rows required for tree building without compromising accuracy<br>
Goss is a kind of sampling technique <br>
<br>
Goss do sampling on the basis of importance of points <br>
This importance of a point is basically absolute value of pseudo resudial (pseudo residual is the differnce between actual value and model's output value till now)<br>
But we can't send only those point who have high pseudo resudual because this will distrub the distribution of the dataset <br>
So what we do is we send top a* total_data_points (0<a=<1) data with high resudal and send some b* total_data_points (0<b=<1) (randomly) from the leftout data to maintain the distribution <br>  
Generally a is greater that b <br>
So due to this we kind of make our model bias for high pseudo resudual data so to avoid this we give each of our data selected from a part weight=1 and other points (selected from b part) to weight = (1-a)/b<br>
by weighting i mean that wherever we will use gradient and hessian we will use them multiplied by weight <br>
now we built tree on these sampled data points and do prediction on whole dataset and then again select pts with highest pseudo residual as done above <br>
top_rate is the hyperparameter for a and other_rate is the hyperparameter for b and in hyperparameter boosting type we tell to apply goss <br>

# EFB
Full form of EFB is exclusive feature bundling<br>
EFB reduces the no of features without compromising accuracy<br>
<br>
it works when we have sparse features <br>
we bundle features together those features which are sparse and not non-zero together <br>
<br>
we do this in two stages<br>
1) Finding out which feature can be bundled together (Greedy Bundling algoritham is used)<br>
2) Bundling selected features (Merge exclusive features algoritham is used)<br>

## Greedy Bundling Algoritham

Firstly it create a max_conflict matrix of m x m (m is the no of features) where each column contain the no of data points where both of the columns have conflict (both have non-zero values together)<br>
after this we create an graph with the above matrix (nodes as feature name / index and matix value's as weight of path)<br>
we choose an threshold and the group of features that have no of conflicts less that threshold we merge them together <br><br>
