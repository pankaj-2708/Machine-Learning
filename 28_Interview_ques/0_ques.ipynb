### 2. Decision Tree (DT) vs. Random Forest (RF): When to Prefer DT

A Random Forest is constructed by combining multiple Decision Trees. While Random Forest generally **outperforms** a single Decision Tree in most cases, there are specific scenarios where DT is preferable.

| Scenario | Explanation | Citation |
| :--- | :--- | :--- |
| **Explainability/Interpretability** | When accuracy is less important than justifying *why* the result was achieved. DT provides a clear **tree-like structure** that can be traced back and plotted to explain the model's workings to stakeholders or clients. | |
| **Computation Power** | RF is a group of DTs, requiring significantly more computation. If working with a large dataset but lacking sufficient computation power, using a single DT can **reduce training time**. | |
| **Feature Importance** | RF randomly selects features. If you are highly confident that certain features are essential for the model, DT is preferred, as RF might skip a crucial feature. | |

### 3. Logistic Regression: Why it is called "Regression"

Although Logistic Regression (LR) is a **classification algorithm**, it is referred to as "regression" because it is a **Generalized Linear Model** that works very similarly to Linear Regression.

*   LR calculates the **probability** of an output (e.g., 0 or 1) using a Sigmoid function.
*   The output probability (e.g., 0.5 or 0.4) is a **continuous value**.
*   Because the output being calculated is this continuous probability value, it is referred to as regression. A **thresholding function** is then applied to convert this continuous probability into a discrete class (0 or 1).

### 4. Out-of-Bag (OOB) Evaluation

OOB stands for **Out-of-Bag evaluation**. This concept is relevant in **bagging** or **Random Forest** algorithms.

*   **Mechanism:** In Random Forest, training involves **sampling with replacement** (bootstrapping). This means some rows (OOB samples) are never selected for inclusion in the training set of a specific Decision Tree.
*   **Usefulness:** Since these samples have **never been seen** by the model, they can be used to **test or validate** the model's performance after training.
*   **Quantity:** Mathematically, approximately **37%** of the data typically remains as OOB samples.

### 5. Naive Bayes: The Reason for "Naive"

Naive Bayes is a supervised machine learning algorithm based on Bayes' Theorem.

*   The term "**Naive**" (meaning "भोला" or innocent) refers to the algorithm's fundamental **assumption**.
*   **The Assumption:** Naive Bayes assumes that all input columns ($X_1, X_2$, etc.) are **conditionally independent** of each other.
*   **Reality:** In real-world datasets, input features usually have some relationship.
*   **Effect:** This simplifying assumption is necessary for the working of Naive Bayes, as it greatly simplifies the calculation of conditional probabilities.

### 6. The No Free Lunch (NFL) Theorem

The No Free Lunch theorem is a significant concept in machine learning, published by computer scientist David Wolpert in 1996.

*   **Core Idea:** If you apply a machine learning model without making *any* assumptions about the nature of the data or the problem, **there is no way to determine in advance** which model will perform best.
*   **Implication:** Machine learning models rely on assumptions (e.g., linear models assume data is linear; Neural Networks assume complex relationships). These assumptions are essential because they save time and prevent the need to run every possible algorithm on every problem.

### 7. Semi-Supervised Machine Learning

Semi-Supervised ML is considered a fourth major type of machine learning, being **partially supervised and partially unsupervised**.

*   **Motivation:** Creating **labels** (the output column) is often costly, difficult, and requires human effort/money.
*   **Technique:** It uses a small amount of **labeled data** to automatically label a large amount of unlabeled data.
*   **Example:** In Google Photos, the system automatically groups photos of the same individual (unsupervised clustering). When a user labels just one photo in a group (e.g., "Mom"), the system automatically labels all other photos in that group, reducing human labour.

### 8. The Unreasonable Effectiveness of Data

This concept arose from experiments conducted by Microsoft researchers Michele Banko and Eric Brill on a Natural Language Processing (NLP) task (Machine Translation).

*   **Discovery:** They found that as the **number of training samples increased**, the performance of all algorithms—even simple ones—began to **converge**, matching the performance of complex algorithms.
*   **Conclusion:** This suggests that the trade-off between **Algorithm Development** (spending time and money creating complex algorithms) and **Dataset Acquisition** (acquiring more data) should be **reconsidered**. In many cases, increasing the size of the dataset may yield results similar to those obtained from highly complex algorithms.

### 9. Mean vs. Median: When to Prefer Median

Both Mean (average) and Median (middle value) are measures of **central tendency**.

*   **Preference for Median:** The median should be preferred when the data contains **outliers**.
*   **Reasoning:** An outlier (an extremely high or low value) significantly skews the **Mean**, leading to a value that does not accurately represent the typical data point. The Median, derived from the sorted middle value, is resistant to the influence of outliers.

### 10. Out-of-Core (OOC) Machine Learning

Out-of-Core machine learning addresses the challenge of working with a dataset that is **larger than the available RAM** (Core Memory) of the machine.

*   **Definition:** OOC ML is a way to train models on data that cannot fit into memory.
*   **OOC Process:**
    1.  **Data Ingestion:** The data is read in **small chunks** from external storage (hard drive) into RAM using a specialised data reader/connector (e.g., libraries like Dask or Spark/PySpark provide this functionality).
    2.  **Feature Extraction:** Features are extracted from these chunks.
    3.  **Training (Incremental Learning):** The model is trained iteratively using **Incremental Learning**. This involves passing mini-batches of data to the model and updating it step-by-step.
*   **Requirement:** This technique requires machine learning algorithms that support a `partial_fit` method. Examples include **SGDClassifier**, **Naive Bayes**, and **Stochastic Gradient Regression**.

### 11. Bagging vs. Boosting (Key Differences)

Bagging and Boosting are ensemble techniques used to build models with low bias and low variance.

| Feature | Bagging (e.g., Random Forest) | Boosting (e.g., XGBoost, AdaBoost) | Citation |
| :--- | :--- | :--- | :--- |
| **Base Model** | Low bias, high variance (e.g., **fully grown Decision Tree**). | High bias, low variance (e.g., **shallow Decision Tree** or stump). | |
| **Training Style** | **Parallel**: All base models are trained independently and simultaneously. | **Sequential**: Models are trained one after another, where later models correct the errors of preceding models. | |
| **Base Learner Weight** | **Equal weight**: All base models have an equal say in the final prediction (like a democracy). | **Unequal weight**: Models that perform better are given a higher weight during the final prediction. | |

### 13. Deciding When to Choose Deep Learning (DL)

Deep Learning is a subset of Machine Learning. The decision to use DL relies on balancing factors that favour DL against those that argue against it.

#### In Favour of Deep Learning:

1.  **High Performance:** DL generally offers better performance, especially as data volume increases, whereas traditional ML model performance often plateaus. DL is preferred for projects where tiny improvements in accuracy (e.g., 0.5%) are critical (e.g., medical diagnoses).
2.  **Complex Problems/Feature Generation:** DL is suited for **complex, open-ended data** like images or text. In these cases, it is difficult or impossible for a human to **hand-craft rules or features**. DL models automatically generate these necessary features.

#### Against Deep Learning (In Favour of Traditional ML):

1.  **Data Requirements:** DL is data-hungry and requires a **large volume of data**, especially **labeled data**. If only a small dataset is available, traditional ML often yields better results.
2.  **Cost/Hardware:** Training DL models usually requires expensive and complex hardware, specifically **GPUs** (Graphics Processing Units), which increase project costs compared to ML.
3.  **Lack of Interpretability:** DL models are often black boxes. They do not provide an answer to **"Why?"** (e.g., why a classification decision was made). Since features are generated internally and lost, there is no way to trace the decision back. If project requirements mandate high interpretability (e.g., justifying a user ban), DL should be avoided.

### 14. Lazy vs. Eager Learning

Machine learning algorithms can be broadly categorised into two types based on when they generalise the training data.

| Feature | Lazy Learning (e.g., K-Nearest Neighbors) | Eager Learning (e.g., Decision Tree, Naive Bayes) | Citation |

### 15. Parametric vs. Non-Parametric Algorithms

Machine learning algorithms are mathematical functions defined by parameters.

#### Parametric Algorithms:

*   **Assumptions:** They **make assumptions** about the functional form of the data (e.g., Linear Regression assumes the data relationship is a line).
*   **Parameters:** The **number of parameters is fixed** and **does not grow** relative to the number of rows in the training data.
*   **Data:** Requires **less data** to train.
*   **Risk:** Prone to **underfitting** if the initial functional assumption is wrong.
*   **Examples:** Linear Regression, Logistic Regression, Naive Bayes, Simple Neural Networks.

#### Non-Parametric Algorithms:

*   **Assumptions:** They **make no explicit assumptions** about the functional form.
*   **Parameters:** The number of parameters or the complexity of the function **grows** with the number of rows (data size).
*   **Data:** Requires **more data** to ensure accurate representation.
*   **Risk:** Prone to **overfitting**.
*   **Examples:** Decision Tree, Random Forest, K-Nearest Neighbors (KNN).

### 16. False Positive (FP) vs. False Negative (FN)

These terms describe incorrect predictions in a binary classification setup.

*   **False Positive (FP):** The model predicts positive (1), but the actual nature is negative (0). This is a **Type 1 Error**.
*   **False Negative (FN):** The model predicts negative (0), but the actual nature is positive (1). This is a **Type 2 Error**.

#### Scenarios for Importance:

1.  **When FP is more important than FN:** **Spam Classification**.
    *   An FP occurs when a legitimate, non-spam email is incorrectly classified as spam. This is serious because the user might miss a crucial email (e.g., a job offer).
    *   An FN occurs when spam is incorrectly classified as non-spam; this is less severe, as the user can often delete the spam manually.

2.  **When FN is more important than FP:** **Cancer Detection**.
    *   An **FN** occurs when a patient *has* cancer but is predicted *not* to have it. This is life-threatening, as the patient will not receive care, and the disease will progress.
    *   An FP occurs when a patient *does not* have cancer but is incorrectly diagnosed with it. This is less severe, as subsequent tests would likely reveal the correct diagnosis.

### 17. Convex vs. Non-Convex Loss Functions

A **Loss Function** is a mathematical measure of the error between predicted and true values, helping to guide model optimisation. Optimization seeks to find the parameter values that **minimise** this loss.

### 18. Online Machine Learning (OML)

Online Machine Learning is a concept used to build products that become more intelligent over time.

*   **Traditional (Batch) ML:** Training is done **offline** using all data simultaneously. Re-training requires the model to be taken down, re-trained offline using new data, and then re-uploaded.
*   **Online ML:** Training is done **online** (on the server). It leverages **Incremental Learning** methods. As users generate new data, the model is continuously and partially trained using **mini-batches** of this new data. The model improves gradually without requiring full, offline re-training.

#### Ideal Scenarios for OML:

1.  **Concept Drift:** When the problem statement or user preferences are expected to **change over time** (e.g., seasonal buying trends on e-commerce sites change the necessary recommendations). OML automatically adjusts the model based on the new concept.
2.  **Cost and Speed:** OML is often **less costly** and training is **much faster** because it works on small, incremental batches of data rather than requiring the entire dataset for retraining.
3.  **Personalisation:** It enables highly **personalised training** based on unique interaction points (e.g., chatbots, mobile keyboards like SwiftKey, or YouTube recommendations adjusting instantly after viewing a video).
