# Optimisations in XGBOOST

in trees for each and every node we try each possible feature to get the best split this technique is also known as Basic Exact Greedy Algorithm<br>
advantage of this - we will get the best split<br>
disadvantage of this - more time complexity<br>

<br>
<br>

Level of Optimisations in XGBOOST<br>
1) Basic Approx Greedy Algorithm
    instead of trying every possible split value for each feature for each node we divide each feature into quantiles and then for each feature we try every quantile value for each node <br>
   so even if at max 100 quantiles is used still we reduce very much calulations  <br>
2) Each quantiles Gradient and hessians are calculated and stored because they are required multiple times in the calculation of similarity score . Max 100 quantiles so only max to max 100 *2 unit memory is used<br>
3) This Gradient and hessians is calculated parallely.<br>
4) This Gradient and hessians value is stored in cache for faster access (as their size is small so it is possible)<br>
5) Quantile Sketch <br>
when data set is to large to load in one computer then we load it into multiple computer then it is not possible to calculate exact quantile because it requires sorting which is not possible (because data cant be loaded into one computer) so we use and algo known as quantile sketch which calculate approx quantile.<br>
6) Weighted Quantlile sketch <br>
   instead of using quantile sketch we use weighted quantile sketch weighted quantile sketch means in each bin instead of no of values sum of weights is same . this weight is same as hessian of the node (hessian is higher for more important node (in classification discussed in previous noteboook) in reg it will be same as hessian is 1 so it weighted quantile binning makes no sense).
