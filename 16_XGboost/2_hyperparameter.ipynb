__*For regularisation in xgboost loss function we add term other than loss function to objective function (check mathematics notebook) . By that we apply reguralisation to each tree in xgboost and in that term parameters are lambda and gamma .*__ 

__There are four ways by which we can regularize our xgboost model.__ <br>
*1) Controlling no of estimators.* <br>
*2) Creating simpler trees.* <br>
*3) By doing sampling (as we do in Random forest).* <br>
*4) By controlling tree height.*

__Above methods in detail.__ <br>

*1) Controlling no of estimators.* <br>
    1) no of estimators <br>
    can be controlled using n_estimators hyperparameter in xgboost <br>
    <br>
    <br>
    &nbsp; 2) early stopping <br>
    &nbsp; in xgboost hyperparameter early_stopping_rounds we can give an integer . xgboost will track loss function value and if after n consecutive (value of early_stopping_rounds ) new trees its value dont decrease  then new tree formation will stop . <br>
    <br>
    <br>
*2) Creating simpler trees.* <br>
    &nbsp; 1) gamma <br>
    &nbsp; Pruning - remove tree nodes<br>
    &nbsp; <img src="../images/x.png">
        <br>
    &nbsp; this is Gain formula<br>
        <br>
    &nbsp; after creating whole tree we revisit tree and check if gain of all nodes below a node is less than 0 then we remove that node and below node
        <br>
    &nbsp; by this way gamma help in regularization <br>
    &nbsp; higher gamma - more prunning -> more regularised <br>
    &nbsp; it can be controlled by gamma hyperparameter in xgboost<br>
    &nbsp; __code for checking gain at each node__ <br>
    ` model_reg.get_booster().get_dump(dump_format='text',with_stats=True)`<br><br>
    by gamma we do post prunning(pruning after tree creation)
    <br>
    <br>
    &nbsp; 2)max_depth <br>
    &nbsp; by this we do pre prunning<br>
    <br>
    <br>
    &nbsp; cover of a node = sum of hessians for all data points in a node  
    &nbsp; for regression (mse) hessian=1
    &nbsp; for classification (log loss) hessian=p(1-p) where p is the probab of previous tree for same pt
    <br>
    <br>
    <br>&nbsp;3) min_child_weight
    <br>&nbsp; min_child_weight is the minimum cover required for splitting (similar to min_sample_split in descission tree)
    <br>&nbsp; for regression it is exactly same as the min_sample_split as hessian=1
    <br>&nbsp; min_child_weight increases => simpler tree => overfitting decreases 
    <br>&nbsp; for classification 
    <br>&nbsp; max hessian can be 0.25
    <br>&nbsp; higher the value of hessian more uncertain our model is for the prediction (think) 
    <br>&nbsp; it do pre prunning
    <br>&nbsp; in a way hessian carry information for the new tree that how our model has performed till now
    <br>&nbsp; in a way we can say that if hessian is higher pt is more important (as it is more uncertain)
    <br>&nbsp; in a way we can say that if cover is higher node is more important (as it is more uncertain)
    <br>&nbsp; min_child_weight is the minimum cover required for splitting 
    <br>&nbsp; if cover is low => pts in that node are more certain => node is less important => no further splitting required
    <br>&nbsp; min_child_weight increases => simpler tree => overfitting decreases 
    
*3) By doing sampling (as we do in Random forest).* <br>
    &nbsp; 1) subsample <br>
    &nbsp; for sampling rows
    <br>
    <br>
    &nbsp; 2) colsample<br>
    &nbsp; for sampling columns<br>
    &nbsp; it has three types <br>
    &nbsp; 1) colsample_bytree<br>
    &nbsp; it give sampled features to a tree<br> 
    &nbsp; 2) colsample_bylevel<br>
    &nbsp; it give sampled features to all nodes of a level of tree (eg level=tree height=2)<br>
    &nbsp; 3) colsample_bynode<br>
    &nbsp; it give sampled features to each node<br>

*4) Shrinkage* <br>
        &nbsp; 1) learning_rate <br>
        &nbsp; can be controlled using eta hyperparameter
        <br>
        &nbsp; 2) lambda <br>
        &nbsp; lambda comes in the denominator in the formula of similarity score and in the denominator of output of a leaf node <br>
        &nbsp; so if we increase lambda -> similarity score decreases -> prunning increases -> overfitting reduces / underfitting increases <br>
        &nbsp; so if we increase lambda -> output of tree decreases (same as learning rate)<br>
        &nbsp; can be controlled using reg_lamda hyperparameter <br>
        &nbsp; both gamma and lambda do prunning so why do we even need lambda? <br>
        &nbsp; if we dont use lambda then we have to set very high values of gamma and because it also control tree output
        <br>
        <br>
        &nbsp; 3) alpha <br>
        &nbsp; <img src='../images/xg.png'>
        &nbsp; here orignally lambda is introduced <br>
        &nbsp; it is like ridge regularisation (sum of square of coefficent is used) <br>
        &nbsp; instead of this we can use lasso regularisation (sum of absolute(coefficient) will be used )<br>
        &nbsp; for controlling it reg_alpha is the hyperparamter <br>
