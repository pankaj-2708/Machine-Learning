# ACF/PACF plots

### DEF- Autocorrelation and partial autocorrelation are statistical measures that help analyze the relationship between a time series and its lagged values.
<img src="https://i.postimg.cc/nrHpNMb7/image.png">
<img src="https://i.postimg.cc/pTgCgkSD/image.png">
<img src="https://i.postimg.cc/W3nr78vN/image.png">

```python
def calculate_acf(X,maxLags):
  X_mean=np.mean(X)
  acf=np.zeros(maxLags)

  X_mean_diff=X-X_mean
  den=sum([(X[i]-X_mean)**2 for i in range(len(X))])
  # acf[0] is one by def
  acf[0]=1
  for i in range(1,maxLags):
    pairs_diff_prod_sum=0
    for j in range(len(X)-i):
        pairs_diff_prod_sum+=X_mean_diff[j]*X_mean_diff[j+i]
    acf[i]=pairs_diff_prod_sum/den
  return acf
```

```python
auto_crr=calculate_acf(df['Close'].values,len(df)//2)
plt.bar([i for i in range(len(df)//2)],auto_crr,width=1)
```

```python
# using libs
from statsmodels.graphics.tsaplots import plot_acf
plot_acf(df['Close'],lags=len(df)/2)
```

# PACF
### DEF - Partial autocorrelation removes the influence of intermediate lags, providing a clearer picture of the direct relationship between a variable and its past values. Unlike autocorrelation, partial autocorrelation focuses on the direct correlation at each lag.

###Failures of acf
**ACF can be misleading because a lag-2 correlation might be happening only because lag-1 exists.**

# 📘 Partial Autocorrelation (PACF) — Intuitive Notes

## ✅ What PACF Measures

PACF (Partial Autocorrelation Function) measures the **direct** relationship between a time-series value \(X_t\) and its past value \(X_{t-k}\) **after removing the influence of intermediate lags**.

$$
\phi_k = \text{Corr}(X_t, X_{t-k} \mid X_{t-1}, X_{t-2}, \ldots, X_{t-k+1})
$$

**Meaning:**  
PACF shows the *pure* effect of lag-\(k\) on \(X_t\), ignoring indirect effects.

---

## 🔁 ACF vs PACF

| Concept | ACF | PACF |
|--------|--------|------|
| Measures | Total autocorrelation | Direct autocorrelation |
| Includes indirect effects? | ✅ Yes | ❌ No |
| Best for | Choosing MA(q) | Choosing AR(p) |

Example:  
If lag-2 depends on lag-1 and lag-1 depends on current value:

- **ACF(2)** may be high (indirect effect)
- **PACF(2)** ≈ 0 (no *direct* effect)

---

## ✅ How PACF is Computed (Concept)

To compute $$(\text{PACF}(k))$$:

1. Regress \(X_t\) on \((X_{t-1}, \ldots, X_{t-k+1})\) → get residuals \(e_t\)
2. Regress \(X_{t-k}\) on \((X_{t-1}, \ldots, X_{t-k+1})\) → get residuals \(e_{t-k}\)
3. Compute

\[
\phi_k = \text{Corr}(e_t,\ e_{t-k})
\]

**Residual correlation = partial correlation**

This is what "conditional covariance" means:  
Correlation **after removing influence of other lags**.

---

## ✅ Dummy Dataset Example

$$[
X = [5, 7, 6, 9, 8, 11, 10, 13, 12, 15]
]$$

Expected behavior:

| Lag | PACF Behavior | Meaning |
|-----|--------------|--------|
| 1 | High | Direct relationship |
| 2 | Small | Weak direct effect |
| 3+ | Near zero | No higher-order influence |

Thus → **AR(1)** structure likely.

---

## ✅ Key Formula Meaning

### ACF vs PACF

$$
\text{ACF}(k) = \text{Corr}(X_t, X_{t-k})$$

$$
\text{PACF}(k) = \text{Corr}(X_t, X_{t-k} \mid X_{t-1}, \ldots, X_{t-k+1})
$$

PACF removes chain effects like

$$
X_{t-2} \rightarrow X_{t-1} \rightarrow X_t
$$

---

## 🎯 Summary

| Topic | Meaning |
|------|--------|
ACF | Total autocorrelation |
PACF | Direct autocorrelation |
Use PACF for | Finding AR order \(p\) |
Use ACF for | Finding MA order \(q\) |

> **PACF finds the real influence of each lag, not the illusion created by other lags.**

---

## 📌 ARIMA Tip

- PACF cutoff → **AR(p)**
- ACF cutoff → **MA(q)**

---

## 🚀 Final One-Line Intuition

> PACF answers:  
> **"Does lag \(k\) directly influence the present, or only through other lags?"**
