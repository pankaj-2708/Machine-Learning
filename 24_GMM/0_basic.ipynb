# Calculating value of K (no of gaussian distibuitions)

We have to manage two quantities . - <br>
1) Minimise k (for simpler model)<br>
2) Maximise likilehood (for better fixed gaussian distribution)<br><br>
   <img src='../../images/lkl.png'>
   <br><br>
   Formula for metting above two requirements is <br>
   1) AIC (Akakie Information Criterion ) = 2*k - 2*ln(L)   (L is the likelhood value calculated by above formula)<br>
   2) BIC (Bayesian Information Criterion ) = ln(n)*k - 2*ln(L)   it also penalise total no of data points (n is the total no of data points and L is the likelhood value calculated by above formula)<br>
   Similar to elbow method graph of both AIC or BIC or one of them is drawn and where elbow is achieved that k value is selected <br>
