# Bayseian optimisation

bayseian optimisation provide intelligence to our hyperparametrs tuning process . it analyse the previous results and find out which hyperparameter combination to try next<br>

def - <br>
Bayseian optimisation is a strategy for optimisation of black-box function that are expensive to evalute . It is particularly useful when dealing with functions that lack an analytical expression , are noisy or are costly in terms of computational resources to evalute. <br>

### Connecting with def
we can write our Score (accuracy,r2score etc) as an function of hyperparameter but we dont know the formula,graph (other things) about this function so it is a black box function (now read def)<br> 

### Objective function

this black box function is also known as objective function <br> 
so we will pass hyperparmetrs values to objective function and it will give as accuracy / mse etc which we will optimise (minimise or maximise)<br>
generally optimise means minimise <br>


## Steps of Bayseian optimisation

1) Define an objective function<br>
&nbsp; eg -
``` python
    def objective(hyper_parametrs):
        df=DecisionTreeClassifier(**hyper_parametrs);
        df.fit(X_train,y_train)
        y_pred=df.predict(X_test)
        return -accuracy(y_pred,y_test) # negative as we want to minimise -accuracy => maximise accuracy
```
2) Define search space <br>
&nbsp; eg- max_depth between 1 to 25 etc

3) Randomly sample some points from search space <br>

4) For each sampled points run objective function <br>

5) Prepare the dataset<br>
&nbsp; dataset D = sampled points and corresponding score for each set of points <br>

6) Surrogate function <br>
&nbsp; surrogate function is a function which try to memic the objective function using this above dataset it is used because objective function is expensive . surrogate function also return and confidence percentange (how much percent it is sure about output given by it)<br><br>
&nbsp; generally used surrogate functions - <br><br>
&nbsp; 1) gaussian process     
&nbsp; 2) random forest <br>
&nbsp; 3) TPE (tree structured parzen estimator ) <br>
<br>

7) fit the surrogate model on dataset D <br>

8) approximate the value of objective function using surrogate function by doing prediction on random points <br>

9) Select an Acquisition function <br><br>
&nbsp; acquisition function is the function which tells us on which next hyperparmeter values should we use objective function <br>
&nbsp; acquisition function finds a trade off between exploration (how much new hyperparmeter's values we should try) and exploitaton (value of objective function (surrogate function ) on a point)<br>
&nbsp; exploration is decided by uncertanity given by surogate function and exploitation is given by accurac/r2 score etc by surrogate function <br><br>
&nbsp; generally used acquistion function - <br>
&nbsp; 1) Expected Improvement (EI)<br>
&nbsp; 2) Upper Confidence Bound (UCB)<br>
&nbsp; 3) Probability of Improvement (PI)<br>

10) randomly select 3 points and send them to surrogate fn to get accuracy and uncertainty <br>

11) send these accuracy and uncertainty to acquistion fn which will then decide the best candidate among these <br>

12) send the best candidate to the objective function to get the accuracy <br>

13) append these new hyperparmeter values and accuracy to dataset D <br>

14) retrain surrogate function on new updated dataset D <br>

15) run these steps (10-14) in loop and get best hyperparmeter value <br>
