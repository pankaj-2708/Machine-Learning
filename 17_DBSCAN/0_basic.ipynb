# DBSCAN

DBSCAN stands for Density-Based Spatial Clustering<br>
DBscan is a density based clustering algo

key terms - 

eps (ε): The maximum distance between two samples for them to be considered as part of the same neighborhood

minPts - the minimum no of points in a neighbourhood for a region to be considered as dense region

Core Points - no of pt's in neighbourhood>=minPts

Border Points - no of pt's in neighbourhood < minPts and in its neighbourhood there is at least one core pt

pts - which are not core pts not border pt

Density connected pts - two pts are density connected if we can put them in one cluster<br>
                        we can put two points in one cluster when - <br>
                        (let's say there are n pts between those pts)<br>

                        1- all n pts between them are core points<br>
                        2- dist b/w any of those two adjacent pts is <= eps (ε)

# Advantages and disadvantages

## Advantages

1- Roboust to outliers <br>
2- No need to specify clusters<br>
3- Can find arbitary shaped clusters<br>
4- Only two hyperparameters to tune<br>

## Disadvantages

1-very sensitive to hyperparameters<br>
2-Unable to classify if density is not uniform across the dataset<br>
3-it don't predict<br>

# sklearn implementation

```python
import numpy as np
from sklearn.datasets import make_circles
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
```

```python
X,y=make_circles(n_samples=500,factor=0.5,noise=0.03,random_state=4)
plt.scatter(X[:,0],X[:,1])
```

```python
for i in range(1,6):
    for j in range(1,5):
        db=DBSCAN(eps=float(i)/10,min_samples=j)
        db.fit_predict(X)
        plt.scatter(X[:,0],X[:,1],c=db.labels_)
        plt.title(f'eps : {float(i)/10} min_samples : {j}')
        plt.show()
```
