# Catboost
Catboost is a boosting framework just like xgboost and lightgbm <br>
Catboost is devloped by Yandex <br>
Catboost can handle categorical data without encoding/bow/tfidf vectorizer <br>

## Technical aspects of catboost

1) It can handle categorical data (text also) using ordered target encoding <br>
2) It creates symmetric trees <br>
3) Calculate leaf nodes output using Newtom raphson method <br>
4) It can dynamically figure out learning rate <br>
5) We can pause training in between and start later from there <br>
6) Built in support for cross_val_score and hyperparameter tuning <br>
7) Can handle missing values<br>
8) GPU support and multi core support <br>
9) Uses pool data structure <br>

## Miscellenaus featues of catboost
1) while fitting catboost model we can also pass validataion data so while training at each step it also check score ( loss function value ) on validation set and rememeber the parametrs which gives best result<br>
2) It let us compare models <br>
