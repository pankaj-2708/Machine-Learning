# Agglomarative clustering

Agglomerative clustering is a hierarchical clustering algorithm that follows a bottom-up approach. It begins with each data point as its own cluster and iteratively merges the closest clusters until all points are in a single cluster or a desired number of clusters is reached. 

# Divisive clustering 

Divisive clustering is a hierarchical clustering technique that uses a "top-down" approach. It starts with all data points in a single cluster and recursively splits it into smaller sub-clusters until each data point is in its own individual cluster or a stopping criterion is met. This contrasts with agglomerative clustering, which uses a "bottom-up" approach, merging individual data points into clusters
