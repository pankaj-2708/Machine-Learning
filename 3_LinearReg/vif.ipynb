```python
# sklearn calculation 
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_boston

data = load_boston()
X = pd.DataFrame(data.data, columns=data.feature_names)

def Ri2_for_feature(X, col):
    X_other = X.loc[:, X.columns != col]
    y = X[col]
    model = LinearRegression().fit(X_other, y)
    Ri2 = model.score(X_other, y)   # returns R^2
    return Ri2

# Example for feature 'RM'
Ri2_RM = Ri2_for_feature(X, 'RM')
print("R_RM^2 =", Ri2_RM)
```

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.datasets import load_boston
import statsmodels.api as sm

# Load dataset
data = load_boston()
X = pd.DataFrame(data.data, columns=data.feature_names)

# Add constant (important for intercept term)
X = sm.add_constant(X)

# Calculate VIF for each feature
vif_data = pd.DataFrame()
vif_data["Feature"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

print(vif_data)
```
