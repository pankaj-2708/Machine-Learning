🔸 Types of Voting Ensembles for classification: 
1. Hard Voting (Majority Voting)
Each model votes for a class label.

The class that gets the most votes wins.

Useful when you only care about the final predicted class.

🧠 Example:
If 3 models predict:

Model A: class 0

Model B: class 1

Model C: class 1
→ Final prediction: class 1

2. Soft Voting (Probability Averaging)
Each model outputs probabilities for each class.

The average of the probabilities is taken, and the class with the highest average probability is selected.

Works only if all models can estimate class probabilities.

🧠 Example:

Model	P(Class 0)	P(Class 1)
A	0.4	0.6
B	0.3	0.7
C	0.5	0.5
→ Average for class 0: (0.4+0.3+0.5)/3 = 0.4		
→ Average for class 1: (0.6+0.7+0.5)/3 = 0.6		
→ Final prediction: class 1
