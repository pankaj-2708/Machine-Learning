🔹 How Voting Works in Regression:

Since regression models output continuous numeric values, there's no concept of "majority vote". So we use averaging instead.

1. Simple Averaging

Each model makes a prediction.

The final prediction is the mean of all model outputs.

2. Weighted Averaging

Each model is given a weight based on its performance.

The final prediction is a weighted mean of all outputs.
